```python
def calculate_churn_rate(column):
    result = pd.crosstab(
        df_clean[column],
        df_clean["Churn"],
        normalize="index"
    ) * 100
    
    return result
```

```python
feature_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "PhoneService",
    "MultipleLines",
    "Partner",
    "Dependents"
]

feature_summary = []

for column in feature_columns:
    temp = (
        df_clean.groupby(column)["Churn_Flag"]
        .mean()
        .mul(100)
        .reset_index()
    )
    
    temp["Feature"] = column
    temp.rename(columns={column: "Category", "Churn_Flag": "Churn_Rate"}, inplace=True)
    
    feature_summary.append(temp)

feature_summary = pd.concat(feature_summary, ignore_index=True)

feature_summary = feature_summary[
    ["Feature", "Category", "Churn_Rate"]
]

feature_summary
```

```python
feature_summary.to_csv(
    "../data/processed/feature_churn_analysis.csv",
    index=False
)

print("Feature churn analysis saved successfully!")
```

```python
df_clean["High_Risk_Customer"] = (
    (df_clean["Contract"] == "Month-to-month") &
    (df_clean["tenure"] <= 12) &
    (df_clean["MonthlyCharges"] >= 70) &
    (df_clean["OnlineSecurity"] == "No") &
    (df_clean["TechSupport"] == "No")
)
```

```python
high_risk_count = df_clean["High_Risk_Customer"].sum()

print("High-Risk Customers:", high_risk_count)
print(
    "Percentage of Customers:",
    round((high_risk_count / len(df_clean)) * 100, 2),
    "%"
)
```

```python
high_risk_churn = pd.crosstab(
    df_clean["High_Risk_Customer"],
    df_clean["Churn"],
    normalize="index"
) * 100

print(high_risk_churn)
```

```python
risk_summary = df_clean.groupby("High_Risk_Customer")[
    ["tenure", "MonthlyCharges", "TotalCharges"]
].mean()

print(risk_summary)
```

```python
risk_summary_table = df_clean.groupby("High_Risk_Customer").agg(
    Customers=("Churn_Flag", "count"),
    Churn_Rate=("Churn_Flag", "mean"),
    Average_Tenure=("tenure", "mean"),
    Average_Monthly_Charges=("MonthlyCharges", "mean")
).reset_index()

risk_summary_table["Churn_Rate"] = (
    risk_summary_table["Churn_Rate"] * 100
)

print(risk_summary_table)
```

```python
risk_summary_table.to_csv(
    "../data/processed/high_risk_customer_analysis.csv",
    index=False
)

print("High-risk customer analysis saved successfully!")
```

```python
query = """
SELECT *
FROM customers
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT COUNT(*) AS total_customers
FROM customers;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT COUNT(*) AS churned_customers
FROM customers
WHERE Churn = 'Yes';
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    ROUND(
        100.0 * SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS churn_rate
FROM customers;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Contract,
    COUNT(*) AS customers,
    ROUND(
        100.0 * SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS churn_rate
FROM customers
GROUP BY Contract
ORDER BY churn_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Tenure_Group,
    COUNT(*) AS customers,
    ROUND(
        100.0 * SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS churn_rate
FROM customers
GROUP BY Tenure_Group
ORDER BY churn_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    PaymentMethod,
    COUNT(*) AS customers,
    ROUND(
        100.0 * SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS churn_rate
FROM customers
GROUP BY PaymentMethod
ORDER BY churn_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    InternetService,
    COUNT(*) AS customers,
    ROUND(
        100.0 * SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS churn_rate
FROM customers
GROUP BY InternetService
ORDER BY churn_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Churn,
    COUNT(*) AS customers,
    ROUND(AVG(tenure), 2) AS average_tenure,
    ROUND(AVG(MonthlyCharges), 2) AS average_monthly_charges,
    ROUND(AVG(TotalCharges), 2) AS average_total_charges
FROM customers
GROUP BY Churn;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    OnlineSecurity,
    COUNT(*) AS customers,
    ROUND(
        100.0 * SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS churn_rate
FROM customers
GROUP BY OnlineSecurity
ORDER BY churn_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    High_Risk_Customer,
    COUNT(*) AS customers,
    ROUND(
        100.0 * SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS churn_rate,
    ROUND(AVG(tenure), 2) AS average_tenure,
    ROUND(AVG(MonthlyCharges), 2) AS average_monthly_charges
FROM customers
GROUP BY High_Risk_Customer
ORDER BY High_Risk_Customer DESC;
"""

pd.read_sql_query(query, conn)
```
