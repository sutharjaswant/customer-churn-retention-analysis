# Customer Churn & Retention Analysis

## 1. Project Overview

This project analyzes customer churn in a telecommunications company using Python, Pandas, SQL, and Power BI. The main objective is to understand why customers leave and identify customer groups that may need retention efforts.

## 2. Key Business Insights

### Insight 1: Overall Customer Churn

The overall customer churn rate is approximately **26.54%**. This indicates that about one in four customers left the company in the dataset.

**Recommendation:** The company should regularly monitor churn and identify customers who may be considering leaving.

### Insight 2: Contract Type

Month-to-month customers have a churn rate of **42.71%**, compared with **11.28%** for one-year contracts and **2.85%** for two-year contracts.

**Recommendation:** Offer suitable long-term contract options and benefits to customers who are willing to commit.

### Insight 3: Customer Tenure

Customers with 0–12 months of tenure have a churn rate of **47.68%**, while customers with 49–72 months of tenure have a churn rate of **9.51%**.

**Recommendation:** Improve onboarding, customer support, and early engagement during the first year.

### Insight 4: Payment Method

Customers using electronic checks have a churn rate of **45.29%**, higher than customers using automatic payment methods.

**Recommendation:** Investigate possible payment-related issues and encourage convenient automatic payment options where appropriate.

### Insight 5: Online Security and Technical Support

Customers without online security have a churn rate of **41.78%**, compared with **14.64%** among customers who have online security. Customers without technical support have a churn rate of **41.65%**, compared with **15.20%** among customers who have technical support.

**Recommendation:** Review customer needs and make relevant security and technical support services easier to understand and access.

### Insight 6: High-Risk Customers

The analysis identified **686 high-risk customers**, representing approximately **9.76%** of the dataset. Their churn rate is **73.03%**, compared with **21.56%** for other customers.

**Recommendation:** Prioritize these customers for proactive support and suitable retention offers.

### Insight 7: Monthly Charges and Customer Tenure

Customers who left had average monthly charges of approximately **74.44**, compared with **61.31** for customers who stayed. Customers who left also had a shorter average tenure of approximately **17.98 months**, compared with **37.65 months** for retained customers.

**Recommendation:** Review pricing concerns, communicate plan value clearly, and identify customers who may need more suitable plans.

## 3. Conclusion

The analysis shows that customer churn is associated with contract type, customer tenure, payment method, service availability, and monthly charges. The company can use these findings to prioritize at-risk customers, improve early customer engagement, and develop better retention strategies.

These findings describe patterns in the dataset and do not prove that any single factor directly causes churn.
